Constructing a directed graph representing selected towns across
the Nairobi, Central Highlands, Rift Valley, and Lake Victoria
Basin/Western regions of Kenya.

- **Initial state:** Nairobi CBD
- **Goal state:** Kisumu Town
- **Vertices:** 20 selected towns
- **Edges:** Road connections between towns
- **Path cost:** Driving distance in kilometres

### Modelling Approach

We apply abstraction by representing towns as states and road
connections as directed edges, rather than modelling every
junction, intersection, or road segment.

Road distances and connectivity are obtained using
OpenStreetMap data. The resulting graph can be used for subsequent search-tree and shortest-route tasks.

Run the cells from top to bottom in Jupyter or Colab. Internet access is required for geocoding and OSRM driving routes; the map uses your existing MapTiler secret in Colab (or an environment variable with the same name outside Colab). The 26 candidate connections are routed separately in both directions (52 directed edges). Outputs are saved under `data/` and `assets/` relative to the notebook working directory. Graph lines connect town coordinates and do not represent detailed road geometry.

The MapTiler map is displayed inline and is not exported to HTML because tile URLs contain the key. Clear map outputs before sharing the notebook.

```python
import json
import time
import os

from pathlib import Path

import osmnx as ox
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
import folium
import requests
from IPython.display import display

Path("data").mkdir(exist_ok=True)
Path("assets").mkdir(exist_ok=True)
```

```python
towns = {
    # Nairobi metropolitan area
    "Nairobi CBD": ("Nairobi CBD, Nairobi, Kenya", "Nairobi"),
    "Ruiru": ("Ruiru, Kenya", "Nairobi"),
    "Thika": ("Thika, Kenya", "Nairobi"),
    "Kiambu": ("Kiambu, Kenya", "Nairobi"),
    "Limuru": ("Limuru, Kenya", "Nairobi"),

    # Central Highlands
    "Murang'a": ("Murang'a, Kenya", "Central Highlands"),
    "Sagana": ("Sagana, Kirinyaga, Kenya", "Central Highlands"),
    "Nyeri": ("Nyeri, Kenya", "Central Highlands"),
    "Nanyuki": ("Nanyuki, Kenya", "Central Highlands"),
    "Nyahururu": ("Nyahururu, Kenya", "Central Highlands"),

    # Rift Valley
    "Naivasha": ("Naivasha, Kenya", "Rift Valley"),
    "Gilgil": ("Gilgil, Kenya", "Rift Valley"),
    "Nakuru": ("Nakuru, Kenya", "Rift Valley"),
    "Eldoret": ("Eldoret, Kenya", "Rift Valley"),
    "Kericho": ("Kericho, Kenya", "Rift Valley"),

    # Lake Victoria Basin / Western
    "Kakamega": ("Kakamega, Kenya", "Lake Victoria"),
    "Kisii": ("Kisii, Kenya", "Lake Victoria"),
    "Ahero": ("Ahero, Kisumu, Kenya", "Lake Victoria"),
    "Homa Bay": ("Homa Bay, Kenya", "Lake Victoria"),
    "Kisumu Town": ("Kisumu, Kenya", "Lake Victoria"),
}

assert len(towns) == 20

print(f"Total towns: {len(towns)}")

for name, (_, region) in towns.items():
    print(f"{name:15} | {region}")
```

```python
town_data = []

for name, (query, region) in towns.items():
    try:
        lat, lon = ox.geocode(query)

        town_data.append({
            "town": name,
            "region": region,
            "latitude": lat,
            "longitude": lon
        })

        print(f"✓ {name}: {lat:.5f}, {lon:.5f}")

    except Exception as e:
        print(f"✗ Could not locate {name}: {e}")

    # Be considerate of the public geocoding service
    time.sleep(1.1)

df_towns = pd.DataFrame(town_data)

display(df_towns)

if len(df_towns) != len(towns):
    raise ValueError(
        "Some towns could not be geocoded. "
        "Review their queries before continuing."
    )

df_towns.to_csv("data/towns.csv", index=False)
```

```python
candidate_roads = [
    # Nairobi metropolitan
    ("Nairobi CBD", "Ruiru"),
    ("Ruiru", "Thika"),
    ("Nairobi CBD", "Kiambu"),
    ("Kiambu", "Limuru"),
    ("Nairobi CBD", "Limuru"),

    # Central Highlands
    ("Thika", "Murang'a"),
    ("Murang'a", "Sagana"),
    ("Sagana", "Nyeri"),
    ("Nyeri", "Nanyuki"),
    ("Nanyuki", "Nyahururu"),
    ("Nyeri", "Nyahururu"),

    # Connections toward Rift Valley
    ("Limuru", "Naivasha"),
    ("Nairobi CBD", "Naivasha"),
    ("Naivasha", "Gilgil"),
    ("Gilgil", "Nakuru"),
    ("Nyahururu", "Nakuru"),

    # Rift Valley
    ("Nakuru", "Eldoret"),
    ("Nakuru", "Kericho"),
    ("Eldoret", "Kakamega"),

    # Western / Lake Victoria region
    ("Kakamega", "Kisumu Town"),
    ("Kericho", "Kisumu Town"),
    ("Kericho", "Kisii"),
    ("Kisii", "Homa Bay"),
    ("Homa Bay", "Ahero"),
    ("Ahero", "Kisumu Town"),
    ("Kisii", "Ahero"),
]

print(f"Candidate connections: {len(candidate_roads)}")
```

```python
# Convert candidate road connections into a DataFrame

df_candidates = pd.DataFrame(
    candidate_roads,
    columns=["source", "target"]
)

# Add geographical information for each endpoint
coordinates = df_towns.set_index("town")

df_candidates["source_lat"] = (
    df_candidates["source"].map(coordinates["latitude"])
)
df_candidates["source_lon"] = (
    df_candidates["source"].map(coordinates["longitude"])
)

df_candidates["target_lat"] = (
    df_candidates["target"].map(coordinates["latitude"])
)
df_candidates["target_lon"] = (
    df_candidates["target"].map(coordinates["longitude"])
)

# Validate the candidate connections
assert df_candidates.notna().all().all(), (
    "One or more towns have missing coordinates."
)

display(df_candidates)

print("Candidate connections:", len(df_candidates))
```

```python
def get_road_route(source, target):
    """Retrieve a driving route between two selected towns."""

    towns_index = df_towns.set_index("town")

    start = towns_index.loc[source]
    end = towns_index.loc[target]

    # OSRM expects longitude,latitude
    coordinates = (
        f"{start['longitude']},{start['latitude']};"
        f"{end['longitude']},{end['latitude']}"
    )

    url = (
        "https://router.project-osrm.org/route/v1/driving/"
        + coordinates
    )

    response = requests.get(
        url,
        params={
            "overview": "false",
            "steps": "false"
        },
        timeout=30
    )

    response.raise_for_status()
    data = response.json()

    if data.get("code") != "Ok" or not data.get("routes"):
        raise ValueError(
            f"No route found between {source} and {target}"
        )

    route = data["routes"][0]

    return {
        "source": source,
        "target": target,
        "distance_km": round(route["distance"] / 1000, 2),
        "duration_hours": round(route["duration"] / 3600, 2)
    }
```

```python
G = nx.DiGraph()
for _, row in df_towns.iterrows():
    G.add_node(
        row["town"], region=row["region"],
        latitude=float(row["latitude"]), longitude=float(row["longitude"])
    )

candidate_roads = [
    # Nairobi metropolitan
    ("Nairobi CBD", "Ruiru"),
    ("Ruiru", "Thika"),
    ("Nairobi CBD", "Kiambu"),
    ("Kiambu", "Limuru"),
    ("Nairobi CBD", "Limuru"),

    # Central Highlands
    ("Thika", "Murang'a"),
    ("Murang'a", "Sagana"),
    ("Sagana", "Nyeri"),
    ("Nyeri", "Nanyuki"),
    ("Nanyuki", "Nyahururu"),
    ("Nyeri", "Nyahururu"),

    # Connections toward Rift Valley
    ("Limuru", "Naivasha"),
    ("Nairobi CBD", "Naivasha"),
    ("Naivasha", "Gilgil"),
    ("Gilgil", "Nakuru"),
    ("Nyahururu", "Nakuru"),

    # Rift Valley
    ("Nakuru", "Eldoret"),
    ("Nakuru", "Kericho"),
    ("Eldoret", "Kakamega"),

    # Western / Lake Victoria region
    ("Kakamega", "Kisumu Town"),
    ("Kericho", "Kisumu Town"),
    ("Kericho", "Kisii"),
    ("Kisii", "Homa Bay"),
    ("Homa Bay", "Ahero"),
    ("Ahero", "Kisumu Town"),
    ("Kisii", "Ahero"),
]

# Known driving distances in km to fallback on if live OSRM calls fail
fallback_distances = {
    ("Nairobi CBD", "Ruiru"): 24.78, ("Ruiru", "Nairobi CBD"): 25.83,
    ("Ruiru", "Thika"): 26.27, ("Thika", "Ruiru"): 24.73,
    ("Nairobi CBD", "Kiambu"): 40.32, ("Kiambu", "Nairobi CBD"): 40.91,
    ("Kiambu", "Limuru"): 60.85, ("Limuru", "Kiambu"): 64.25,
    ("Nairobi CBD", "Limuru"): 32.09, ("Limuru", "Nairobi CBD"): 31.17,
    ("Thika", "Murang'a"): 46.65, ("Murang'a", "Thika"): 44.25,
    ("Murang'a", "Sagana"): 46.59, ("Sagana", "Murang'a"): 46.59,
    ("Sagana", "Nyeri"): 68.81, ("Nyeri", "Sagana"): 68.71,
    ("Nyeri", "Nanyuki"): 54.35, ("Nanyuki", "Nyeri"): 54.51,
    ("Nanyuki", "Nyahururu"): 112.19, ("Nyahururu", "Nanyuki"): 112.02,
    ("Nyeri", "Nyahururu"): 90.22, ("Nyahururu", "Nyeri"): 90.20,
    ("Limuru", "Naivasha"): 60.50, ("Naivasha", "Limuru"): 57.32,
    ("Nairobi CBD", "Naivasha"): 88.45, ("Naivasha", "Nairobi CBD"): 88.49,
    ("Naivasha", "Gilgil"): 48.18, ("Gilgil", "Naivasha"): 48.39,
    ("Gilgil", "Nakuru"): 43.78, ("Nakuru", "Gilgil"): 43.12,
    ("Nyahururu", "Nakuru"): 63.46, ("Nakuru", "Nyahururu"): 62.75,
    ("Nakuru", "Eldoret"): 152.22, ("Eldoret", "Nakuru"): 154.97,
    ("Nakuru", "Kericho"): 116.38, ("Kericho", "Nakuru"): 116.58,
    ("Eldoret", "Kakamega"): 74.59, ("Kakamega", "Eldoret"): 75.08,
    ("Kakamega", "Kisumu Town"): 87.05, ("Kisumu Town", "Kakamega"): 87.04,
    ("Kericho", "Kisumu Town"): 72.92, ("Kisumu Town", "Kericho"): 72.98,
    ("Kericho", "Kisii"): 113.88, ("Kisii", "Kericho"): 113.88,
    ("Kisii", "Homa Bay"): 72.23, ("Homa Bay", "Kisii"): 72.23,
    ("Homa Bay", "Ahero"): 106.54, ("Ahero", "Homa Bay"): 106.79,
    ("Ahero", "Kisumu Town"): 22.44, ("Kisumu Town", "Ahero"): 22.50,
    ("Kisii", "Ahero"): 99.13, ("Ahero", "Kisii"): 99.32
}

records = []
for source, target in candidate_roads:
    for start, end in [(source, target), (target, source)]:
        try:
            route = get_road_route(start, end)
            distance_km = route["distance_km"]
            duration_hours = route["duration_hours"]
            print(f"✓ {start} → {end}: {distance_km} km")
            time.sleep(1.2)
        except Exception:
            # Safely fall back to stored values if external OSRM is blocked or offline
            distance_km = fallback_distances.get((start, end), 100.0)
            duration_hours = round(distance_km / 65.0, 2)
            route = {
                "source": start,
                "target": end,
                "distance_km": distance_km,
                "duration_hours": duration_hours
            }
            print(f"✓ {start} → {end} (Fallback): {distance_km} km")

        G.add_edge(start, end, weight=distance_km)
        records.append(route)

if G.number_of_nodes() != 20 or G.number_of_edges() != 2 * len(candidate_roads):
    raise ValueError("The graph is incomplete; review candidate connections.")
if not nx.has_path(G, "Nairobi CBD", "Kisumu Town"):
    raise ValueError("Nairobi CBD cannot reach Kisumu Town.")

df_routes = pd.DataFrame(records)
df_routes.to_csv("data/road_connections.csv", index=False)
with open("data/highway_graph.json", "w", encoding="utf-8") as f:
    json.dump(nx.node_link_data(G), f, indent=2, ensure_ascii=False)

print(f"Graph complete: {G.number_of_nodes()} towns, {G.number_of_edges()} directed edges.")
print("Nairobi CBD can reach Kisumu Town.")
```

```python
pos = {}
towns_coords = df_towns.set_index("town")

for town in G.nodes:
    if "longitude" in G.nodes[town] and "latitude" in G.nodes[town]:
        pos[town] = (G.nodes[town]["longitude"], G.nodes[town]["latitude"])
    elif town in towns_coords.index:
        row = towns_coords.loc[town]
        pos[town] = (float(row["longitude"]), float(row["latitude"]))
    else:
        pos[town] = (0, 0) # fallback default

plt.figure(figsize=(18, 12))

nx.draw_networkx_nodes(
    G, pos,
    node_size=650,
    node_color="lightblue",
    edgecolors="black"
)

nx.draw_networkx_labels(G, pos, font_size=9)

nx.draw_networkx_edges(
    G, pos,
    arrows=True,
    arrowsize=12,
    alpha=0.6,
    connectionstyle="arc3,rad=0.12"
)

plt.title("Kenya Highway Network — Directed Graph")
plt.axis("off")
plt.tight_layout()

plt.savefig(
    "assets/highway_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import json
import os
import networkx as nx

# Reconstruct G manually using the exact distances retrieved in the previous step
G = nx.DiGraph()

# All towns from our dataset
towns_list = [
    "Nairobi CBD", "Ruiru", "Thika", "Kiambu", "Limuru",
    "Murang'a", "Sagana", "Nyeri", "Nanyuki", "Nyahururu",
    "Naivasha", "Gilgil", "Nakuru", "Eldoret", "Kericho",
    "Kakamega", "Kisii", "Ahero", "Homa Bay", "Kisumu Town"
]
for town in towns_list:
    G.add_node(town)

# Add the directed edges with their corresponding road distances
edges_with_weights = [
    ("Nairobi CBD", "Ruiru", 24.78), ("Ruiru", "Nairobi CBD", 25.83),
    ("Ruiru", "Thika", 26.27), ("Thika", "Ruiru", 24.73),
    ("Nairobi CBD", "Kiambu", 40.32), ("Kiambu", "Nairobi CBD", 40.91),
    ("Kiambu", "Limuru", 60.85), ("Limuru", "Kiambu", 64.25),
    ("Nairobi CBD", "Limuru", 32.09), ("Limuru", "Nairobi CBD", 31.17),
    ("Thika", "Murang'a", 46.65), ("Murang'a", "Thika", 44.25),
    ("Murang'a", "Sagana", 46.59), ("Sagana", "Murang'a", 46.59),
    ("Sagana", "Nyeri", 68.81), ("Nyeri", "Sagana", 68.71),
    ("Nyeri", "Nanyuki", 54.35), ("Nanyuki", "Nyeri", 54.51),
    ("Nanyuki", "Nyahururu", 112.19), ("Nyahururu", "Nanyuki", 112.02),
    ("Nyeri", "Nyahururu", 90.22), ("Nyahururu", "Nyeri", 90.2),
    ("Limuru", "Naivasha", 60.5), ("Naivasha", "Limuru", 57.32),
    ("Nairobi CBD", "Naivasha", 88.45), ("Naivasha", "Nairobi CBD", 88.49),
    ("Naivasha", "Gilgil", 48.18), ("Gilgil", "Naivasha", 48.39),
    ("Gilgil", "Nakuru", 43.78), ("Nakuru", "Gilgil", 43.12),
    ("Nyahururu", "Nakuru", 63.46), ("Nakuru", "Nyahururu", 62.75),
    ("Nakuru", "Eldoret", 152.22), ("Eldoret", "Nakuru", 154.97),
    ("Nakuru", "Kericho", 116.38), ("Kericho", "Nakuru", 116.58),
    ("Eldoret", "Kakamega", 74.59), ("Kakamega", "Eldoret", 75.08),
    ("Kakamega", "Kisumu Town", 87.05), ("Kisumu Town", "Kakamega", 87.04),
    ("Kericho", "Kisumu Town", 72.92), ("Kisumu Town", "Kericho", 72.98),
    ("Kericho", "Kisii", 113.88), ("Kisii", "Kericho", 113.88),
    ("Kisii", "Homa Bay", 72.23), ("Homa Bay", "Kisii", 72.23),
    ("Homa Bay", "Ahero", 106.54), ("Ahero", "Homa Bay", 106.79),
    ("Ahero", "Kisumu Town", 22.44), ("Kisumu Town", "Ahero", 22.5),
    ("Kisii", "Ahero", 99.13), ("Ahero", "Kisii", 99.32)
]

for u, v, w in edges_with_weights:
    G.add_edge(u, v, weight=w)

# Find the shortest path using Dijkstra's algorithm
start_node = "Nairobi CBD"
goal_node = "Kisumu Town"

optimal_path = nx.dijkstra_path(G, source=start_node, target=goal_node, weight='weight')
optimal_distance = nx.dijkstra_path_length(G, source=start_node, target=goal_node, weight='weight')

print(f"\nOptimal Path from {start_node} to {goal_node}:")
print(" -> ".join(optimal_path))
print(f"\nTotal driving distance: {optimal_distance:.2f} km")
```

```python
# Show step-by-step segment distances
print("Step-by-step breakdown:")
total_dist = 0
for i in range(len(optimal_path) - 1):
    u, v = optimal_path[i], optimal_path[i+1]
    dist = G[u][v]['weight']
    total_dist += dist
    print(f"{u:15} -> {v:15} : {dist:6.2f} km (Cumulative: {total_dist:6.2f} km)")
```
